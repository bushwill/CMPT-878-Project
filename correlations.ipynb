```python
#imports
import os
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
import helpers
```

Directory/File Name settings¶

```python
dataset_directory = "CamelinaMAGIC-Dataset-Complete/"

MAGIC1_daily_transpiration_file = os.path.join(dataset_directory, "CamelinaMAGIC1.0-DailyTranspiration.csv")
MAGIC1_weight_raw = os.path.join(dataset_directory, "CamelinaMAGIC1.0-WeightRaw.csv")
MAGIC1_weather_station = os.path.join(dataset_directory, "CamelinaMAGIC1.0-WeatherStation.csv")

MAGIC2_daily_transpiration_file = os.path.join(dataset_directory, "CamelinaMAGIC2.0-DailyTranspiration.csv")
MAGIC2_weight_raw = os.path.join(dataset_directory, "CamelinaMAGIC2.0-WeightRaw.csv")
MAGIC2_weather_station = os.path.join(dataset_directory, "CamelinaMAGIC2.0-WeatherStation.csv")

MAGIC3_daily_transpiration_file = os.path.join(dataset_directory, "CamelinaMAGIC3.0-DailyTranspiration.csv")
MAGIC3_weight_raw = os.path.join(dataset_directory, "CamelinaMAGIC3.0-WeightRaw.csv")
MAGIC3_weather_station = os.path.join(dataset_directory, "CamelinaMAGIC3.0-WeatherStation.csv")

MAGIC_randomization_file = os.path.join(dataset_directory, "CamelinaMAGIC-ALL-Randomization.xlsx")

output_directory = "cor_output/"
```

transpiration dataframes

```python
# Load Experiment 1 transpiration data
transpiration_df_1 = helpers.read_transpiration_csv(MAGIC1_daily_transpiration_file, MAGIC_randomization_file, 1)
#reduce trial time to 40 days even across all experiments
transpiration_df_1 = transpiration_df_1[transpiration_df_1['Days'] <= 40]
print(transpiration_df_1)

# Load Experiment 2 transpiration data
transpiration_df_2 = helpers.read_transpiration_csv(MAGIC2_daily_transpiration_file, MAGIC_randomization_file, 2)
#reduce trial time to 40 days even across all experiments
transpiration_df_2 = transpiration_df_2[transpiration_df_2['Days'] <= 40]
print(transpiration_df_2)

# Load Experiment 3 transpiration data
transpiration_df_3 = helpers.read_transpiration_csv(MAGIC3_daily_transpiration_file, MAGIC_randomization_file, 3)
#reduce trial time to 40 days even across all experiments
transpiration_df_3 = transpiration_df_3[transpiration_df_3['Days'] <= 40]
print(transpiration_df_3)
```

climate dataframes

```python
# Load Experiment 1 weather station data
weather_df_1 = helpers.read_weather_station_csv(MAGIC1_weather_station, 1, "24H")
#reduce trial time to 40 days even across all experiments
weather_df_1 = weather_df_1[weather_df_1['Days'] <= 40]
print(weather_df_1)

# Load Experiment 2 weather station data
weather_df_2 = helpers.read_weather_station_csv(MAGIC2_weather_station, 2, "24H")
#reduce trial time to 40 days even across all experiments
weather_df_2 = weather_df_2[weather_df_2['Days'] <= 40]
print(weather_df_2)

# Load Experiment 3 weather station data
weather_df_3 = helpers.read_weather_station_csv(MAGIC3_weather_station, 3, "24H")
#reduce trial time to 40 days even across all experiments
weather_df_3 = weather_df_3[weather_df_3['Days'] <= 40]
print(weather_df_3)
```

merge dataframes for correlation matrix

```python
#experiment1
#inner merge (only matching Days values)
merged_df_1 = pd.merge(
    left=transpiration_df_1,
    right=weather_df_1,
    on=["Days", "Experiment"],
    how="inner"
)

# Check for any missing values that might indicate merge issues
print(f"Original row counts: Transpiration {len(transpiration_df_1)}, Weather {len(weather_df_1)}")
print(f"Merged dataframe row count: {len(merged_df_1)}")
print(f"Missing values in merged dataframe:\n{merged_df_1.isna().sum()}")

# Display the first few rows
merged_df_1.head()

#experiment2
#inner merge (only matching Days values)
merged_df_2 = pd.merge(
    left=transpiration_df_2,
    right=weather_df_2,
    on=["Days", "Experiment"],
    how="inner"
)

# Check for any missing values that might indicate merge issues
print(f"Original row counts: Transpiration {len(transpiration_df_2)}, Weather {len(weather_df_2)}")
print(f"Merged dataframe row count: {len(merged_df_2)}")
print(f"Missing values in merged dataframe:\n{merged_df_2.isna().sum()}")

# Display the first few rows
merged_df_2.head()

#experiment3
#inner merge (only matching Days values)
merged_df_3 = pd.merge(
    left=transpiration_df_3,
    right=weather_df_3,
    on=["Days", "Experiment"],
    how="inner"
)

# Check for any missing values that might indicate merge issues
print(f"Original row counts: Transpiration {len(transpiration_df_3)}, Weather {len(weather_df_3)}")
print(f"Merged dataframe row count: {len(merged_df_3)}")
print(f"Missing values in merged dataframe:\n{merged_df_3.isna().sum()}")

# Display the first few rows
merged_df_3.head()

#merge all
# Concatenate the three dataframes vertically (stacking rows)
merged_df_ALL = pd.concat([merged_df_1, merged_df_2, merged_df_3], 
                        axis=0,  # 0 means stack vertically (rows)
                        ignore_index=True)  # Reset the index to avoid duplicates

# Display information about the original and combined dataframes
print(f"Original shapes:")
print(f"merged_df_1: {merged_df_1.shape}")
print(f"merged_df_2: {merged_df_2.shape}")
print(f"merged_df_3: {merged_df_3.shape}")
print(f"Combined dataframe shape: {merged_df_ALL.shape}")

# Check if the row count adds up correctly
expected_rows = len(merged_df_1) + len(merged_df_2) + len(merged_df_3)
actual_rows = len(merged_df_ALL)
print(f"Expected rows: {expected_rows}, Actual rows: {actual_rows}")

# Display the first few rows of the combined dataframe
merged_df_ALL.head(1020)
```

clean transpiration data

```python
# Store the original shape for comparison
original_shape = merged_df_ALL.shape

# Check for any extreme values before filtering
print("Before filtering:")
print(f"Min Transpiration value: {merged_df_ALL['Transpiration'].min()}")
print(f"Max Transpiration value: {merged_df_ALL['Transpiration'].max()}")
print(f"Number of rows with Transpiration < 0: {(merged_df_ALL['Transpiration'] < 0).sum()}")
print(f"Number of rows with Transpiration > 500: {(merged_df_ALL['Transpiration'] > 500).sum()}")

# Filter the dataframe to keep only rows where Transpiration is between 0 and 500 (inclusive)
filtered_df = merged_df_ALL[(merged_df_ALL['Transpiration'] >= 0) & (merged_df_ALL['Transpiration'] <= 500)]

# Reset the index if needed
filtered_df = filtered_df.reset_index(drop=True)

# Check the results
print("\nAfter filtering:")
print(f"Original shape: {original_shape}")
print(f"New shape: {filtered_df.shape}")
print(f"Removed {original_shape[0] - filtered_df.shape[0]} rows")

# Verify the new range
print(f"New Min Transpiration value: {filtered_df['Transpiration'].min()}")
print(f"New Max Transpiration value: {filtered_df['Transpiration'].max()}")
```

```python
from scipy.stats import pearsonr, spearmanr, pointbiserialr
```

```python
#correlation matrix (excludes genotype)

# Create a copy of dataframe for analysis
analysis_df = transform_d2.copy()

# 1. Handle the binary Treatment variable - convert to numeric (0/1)
analysis_df['Treatment_numeric'] = (analysis_df['Treatment'] == 'DR_100').astype(int)
# Now 0 = Control, 1 = DR_100

# 2. Select numerical columns for correlation analysis
# Include original numeric columns and Treatment_numeric
numeric_cols = analysis_df.select_dtypes(include=[np.number]).columns
correlation_df = analysis_df[numeric_cols]

# 3. Calculate and visualize the correlation matrix
correlation_matrix = correlation_df.corr()

# Create a visualization
plt.figure(figsize=(14, 12))
sns.heatmap(correlation_matrix, 
            annot=False,          # Too many values to show annotations
            cmap='coolwarm',
            vmin=-1, vmax=1,
            linewidths=0.5)

plt.title('Correlation Matrix', fontsize=16)
plt.tight_layout()
plt.show()
```

4. plot specific relationships

binary vs continuous

```python
# Treatment vs Transpiration
if 'Transpiration' in correlation_df.columns:
    # Point-biserial correlation for binary vs continuous
    corr, p_value = pointbiserialr(analysis_df['Treatment_numeric'], analysis_df['Transpiration'])
    print(f"Correlation between Treatment and Transpiration: {corr:.3f} (p-value: {p_value:.4f})")
    
    # Boxplot to visualize the relationship
    plt.figure(figsize=(10, 6))
    sns.boxplot(x='Treatment', y='Transpiration', data=analysis_df)
    plt.title(f'Distribution of Transpiration by Treatment')
    plt.show()
```

```python
# Experiment vs Transpiration
if 'Transpiration' in correlation_df.columns:
    # Point-biserial correlation for binary vs continuous
    corr, p_value = pointbiserialr(analysis_df['Experiment'], analysis_df['Transpiration'])
    print(f"Correlation between Experiment and Transpiration: {corr:.3f} (p-value: {p_value:.4f})")
    
    # Boxplot to visualize the relationship
    plt.figure(figsize=(10, 6))
    sns.boxplot(x='Experiment', y='Transpiration', data=analysis_df)
    plt.title(f'Distribution of Transpiration by Experiment')
    plt.show()
```

continuous vs continuous

```python
# Days vs Transpiration
resp_var = 'Transpiration'
exp_var_1 = 'Days'

if resp_var in analysis_df.columns and exp_var_1 in analysis_df.columns:
    # Pearson correlation for continuous vs continuous variables
    pearson_corr, pearson_p_value = pearsonr(analysis_df[exp_var_1], analysis_df[resp_var])
    print(f"Pearson correlation between {exp_var_1} and {resp_var}: {pearson_corr:.3f} (p-value: {pearson_p_value:.4f})")
    
    # Optional: Also calculate Spearman rank correlation (useful if relationship might be non-linear)
    spearman_corr, spearman_p_value = spearmanr(analysis_df[exp_var_1], analysis_df[resp_var])
    print(f"Spearman correlation between {exp_var_1} and {resp_var}: {spearman_corr:.3f} (p-value: {spearman_p_value:.4f})")
    
    # Interpret the correlation strength
    if abs(pearson_corr) < 0.3:
        strength = "weak"
    elif abs(pearson_corr) < 0.7:
        strength = "moderate"
    else:
        strength = "strong"
    
    direction = "positive" if pearson_corr > 0 else "negative"
    
    print(f"Interpretation: There is a {strength} {direction} correlation between {exp_var_1} and {resp_var}.")
    
    # If p-value is significant
    if spearman_p_value < 0.05:
        print(f"This correlation (Spearman) is statistically significant (p < 0.05).")
    else:
        print(f"This correlation (Spearman) is not statistically significant (p > 0.05).")
else:
    print("Required columns '{exp_var_1}' or '{resp_var}' not found in the dataframe.")
```

```python
anova_df = filtered_df.copy()

# 1.  visualize the data
plt.figure(figsize=(12, 6))
sns.boxplot(x='Genotype', y='Transpiration', hue='Treatment', data=anova_df)
plt.title('Transpiration by Genotype and Treatment')
plt.xlabel('Genotype')
plt.ylabel('Transpiration')
plt.xticks(rotation=45)  # Rotate x labels if needed
plt.legend(title='Treatment')
plt.tight_layout()
plt.show()

# Optional:  interaction plot to visualize potential interactions
plt.figure(figsize=(10, 6))
sns.pointplot(x='Genotype', y='Transpiration', hue='Treatment', data=anova_df, dodge=0.5, ci=95)
plt.title('Interaction Plot: Genotype × Treatment')
plt.xlabel('Genotype')
plt.ylabel('Mean Transpiration')
plt.xticks(rotation=45)
plt.legend(title='Treatment')
plt.tight_layout()
plt.show()

# 2. Run the two-way ANOVA using statsmodels
# Create the model formula with main effects and interaction term
model = ols('Transpiration ~ C(Genotype) + C(Treatment) + C(Genotype):C(Treatment)', data=anova_df).fit()

# Create the ANOVA table (Type II SS is generally recommended for unbalanced designs)
anova_table = sm.stats.anova_lm(model, typ=2)

# Display the ANOVA table
print("Two-way ANOVA Table:")
print(anova_table)

# 3. Interpret the results
print("\nInterpretation:")

# Main effect of Genotype
p_genotype = anova_table['PR(>F)'][0]
if p_genotype < 0.05:
    print(f"- There is a statistically significant main effect of Genotype (p-value: {p_genotype:.4f}).")
else:
    print(f"- There is no statistically significant main effect of Genotype (p-value: {p_genotype:.4f}).")

# Main effect of Treatment
p_treatment = anova_table['PR(>F)'][1]
if p_treatment < 0.05:
    print(f"- There is a statistically significant main effect of Treatment (p-value: {p_treatment:.4f}).")
else:
    print(f"- There is no statistically significant main effect of Treatment (p-value: {p_treatment:.4f}).")

# Interaction effect
p_interaction = anova_table['PR(>F)'][2]
if p_interaction < 0.05:
    print(f"- There is a statistically significant interaction between Genotype and Treatment (p-value: {p_interaction:.4f}).")
    print("  This means the effect of Treatment on Transpiration depends on the Genotype (and vice versa).")
else:
    print(f"- There is no statistically significant interaction between Genotype and Treatment (p-value: {p_interaction:.4f}).")
    print("  This means the effect of Treatment on Transpiration is consistent across all Genotypes.")

# 4. Post-hoc tests if main effects or interactions are significant
if p_genotype < 0.05:
    print("\nPost-hoc test for Genotype main effect:")
    tukey_genotype = pairwise_tukeyhsd(endog=anova_df['Transpiration'], 
                                      groups=anova_df['Genotype'],
                                      alpha=0.05)
    print(tukey_genotype)

if p_treatment < 0.05:
    print("\nPost-hoc test for Treatment main effect:")
    tukey_treatment = pairwise_tukeyhsd(endog=anova_df['Transpiration'], 
                                       groups=anova_df['Treatment'],
                                       alpha=0.05)
    print(tukey_treatment)

if p_interaction < 0.05:
    print("\nSince there is a significant interaction, you should analyze simple effects.")
    print("This means examining the effect of one factor at each level of the other factor.")
    
    # Example of simple effects analysis - examining Treatment effect within each Genotype
    print("\nSimple effects analysis (Treatment effects within each Genotype):")
    genotypes = anova_df['Genotype'].unique()
    
    for genotype in genotypes:
        subset = anova_df[anova_df['Genotype'] == genotype]
        model_simple = ols('Transpiration ~ C(Treatment)', data=subset).fit()
        anova_simple = sm.stats.anova_lm(model_simple, typ=2)
        
        print(f"\nEffect of Treatment within Genotype '{genotype}':")
        print(anova_simple)
        
        # If significant, run Tukey's test
        if anova_simple['PR(>F)'][0] < 0.05:
            tukey_simple = pairwise_tukeyhsd(endog=subset['Transpiration'], 
                                           groups=subset['Treatment'],
                                           alpha=0.05)
            print(f"Tukey's HSD for Treatment within Genotype '{genotype}':")
            print(tukey_simple)
```

```python
import numpy as np
import pandas as pd
import statsmodels.api as sm
import matplotlib.pyplot as plt
import seaborn as sns

nonNeg_norm_df = filtered_df.copy()

# Method 1: Simple ratio - Transpiration per unit VPD
nonNeg_norm_df['Transpiration_per_VPD'] = nonNeg_norm_df['Transpiration'] / nonNeg_norm_df['VPD']
# Handle potential division by zero
nonNeg_norm_df['Transpiration_per_VPD'] = nonNeg_norm_df['Transpiration_per_VPD'].replace([np.inf, -np.inf], np.nan)

#shift so no negative transpiration values
min_val1 = nonNeg_norm_df['Transpiration_per_VPD'].min()
if min_val1 < 0:
    nonNeg_norm_df['Transpiration_per_VPD'] = nonNeg_norm_df['Transpiration_per_VPD'] - min_val1

# Method 2: Linear regression-based normalization
X = sm.add_constant(nonNeg_norm_df['VPD'])
model = sm.OLS(nonNeg_norm_df['Transpiration'], X).fit()

# Residuals (Transpiration after removing VPD effect)
nonNeg_norm_df['Transpiration_VPD_adjusted'] = model.resid

#shift so no negative transpiration values
min_val2 = nonNeg_norm_df['Transpiration_VPD_adjusted'].min()
if min_val2 < 0:
    nonNeg_norm_df['Transpiration_VPD_adjusted'] = nonNeg_norm_df['Transpiration_VPD_adjusted'] - min_val2

# Method 3: ANCOVA-like approach (group-specific)
nonNeg_norm_df['Transpiration_VPD_adjusted_by_group'] = np.nan

for group in nonNeg_norm_df['Experiment'].unique():
    group_data = nonNeg_norm_df[nonNeg_norm_df['Experiment'] == group]
    X = sm.add_constant(group_data['VPD'])
    try:
        group_model = sm.OLS(group_data['Transpiration'], X).fit()
        # use .loc with group_data.index
        nonNeg_norm_df.loc[group_data.index, 'Transpiration_VPD_adjusted_by_group'] = group_model.resid
    except:
        print(f"Could not fit model for group {group}")

#shift so no negative transpiration values
min_val3 = nonNeg_norm_df['Transpiration_VPD_adjusted_by_group'].min()
if min_val3 < 0:
    nonNeg_norm_df['Transpiration_VPD_adjusted_by_group'] = nonNeg_norm_df['Transpiration_VPD_adjusted_by_group'] - min_val3

# Method 4: Standardized residuals
nonNeg_norm_df['Transpiration_VPD_standardized_residuals'] = model.get_influence().resid_studentized_internal

#shift so no negative transpiration values
min_val4 = nonNeg_norm_df['Transpiration_VPD_standardized_residuals'].min()
if min_val4 < 0:
    nonNeg_norm_df['Transpiration_VPD_standardized_residuals'] = nonNeg_norm_df['Transpiration_VPD_standardized_residuals'] - min_val4

# Method 5: Reference VPD normalization
reference_VPD = nonNeg_norm_df['VPD'].mean()
predicted_at_reference = model.params[0] + model.params[1] * reference_VPD
nonNeg_norm_df['Transpiration_at_reference_VPD'] = predicted_at_reference + model.resid

min_val5 = nonNeg_norm_df['Transpiration_at_reference_VPD'].min()
if min_val5 < 0:
    nonNeg_norm_df['Transpiration_at_reference_VPD'] = nonNeg_norm_df['Transpiration_at_reference_VPD'] - min_val5


# Visualizations
plt.figure(figsize=(15, 10))

# Original relationship
plt.subplot(2, 3, 1)
sns.scatterplot(x='VPD', y='Transpiration', data=nonNeg_norm_df)
plt.title('Original: Transpiration vs VPD')

# Simple ratio
plt.subplot(2, 3, 2)
sns.scatterplot(x='VPD', y='Transpiration_per_VPD', data=nonNeg_norm_df)
plt.title('Transpiration per unit VPD')

# Regression-adjusted
plt.subplot(2, 3, 3)
sns.scatterplot(x='VPD', y='Transpiration_VPD_adjusted', data=nonNeg_norm_df)
plt.title('Regression-adjusted Transpiration')
plt.axhline(y=0, color='r', linestyle='-', alpha=0.3)

# Group-specific adjustment
plt.subplot(2, 3, 4)
sns.scatterplot(x='VPD', y='Transpiration_VPD_adjusted_by_group', 
                hue='Experiment', data=nonNeg_norm_df)
plt.title('Group-specific VPD Adjustment')
plt.axhline(y=0, color='r', linestyle='-', alpha=0.3)

# Standardized residuals
plt.subplot(2, 3, 5)
sns.scatterplot(x='VPD', y='Transpiration_VPD_standardized_residuals', data=nonNeg_norm_df)
plt.title('Standardized Residuals')
plt.axhline(y=0, color='r', linestyle='-', alpha=0.3)

# Reference VPD normalization
plt.subplot(2, 3, 6)
sns.scatterplot(x='VPD', y='Transpiration_at_reference_VPD', data=nonNeg_norm_df)
plt.title(f'Transpiration at Reference VPD ({reference_VPD:.2f})')

plt.tight_layout()
plt.show()

# Before vs After normalization
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

sns.regplot(x='VPD', y='Transpiration', data=nonNeg_norm_df, ax=ax1)
ax1.set_title('Before Normalization')
ax1.set_xlabel('VPD')
ax1.set_ylabel('Transpiration')

sns.regplot(x='VPD', y='Transpiration_VPD_adjusted', data=nonNeg_norm_df, ax=ax2)
ax2.set_title('After Normalization')
ax2.set_xlabel('VPD')
ax2.set_ylabel('VPD-adjusted Transpiration')
ax2.axhline(y=0, color='r', linestyle='-', alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
##find days with different mean of each parameter per experiment

from scipy import stats

# Function to calculate mean transpiration per genotype per treatment
def calculate_mean_transpiration(df):
    # Group by Day, Genotype, and Treatment, then calculate mean
    mean_transpiration = df.groupby(['Days', 'Genotype', 'Treatment'])['Transpiration'].mean().reset_index()
    # Pivot to have days as index, and columns as Genotype_Treatment combinations
    means_df = mean_transpiration.pivot_table(
        index='Days', 
        columns=['Genotype', 'Treatment'], 
        values='Transpiration'
    ).sort_index()
    # Calculate day-to-day differences
    diffs = means_df.diff().dropna()
    
    return means_df, diffs

means_1, diffs_1 = calculate_mean_transpiration(transpiration_df_1)
means_2, diffs_2 = calculate_mean_transpiration(transpiration_df_2)
means_3, diffs_3 = calculate_mean_transpiration(transpiration_df_3)

# Function to detect significant deviations
def detect_significant_days(diffs):
    significant_days = {}
    mean_changes = {}
    
    for column in diffs.columns:
        # Calculate mean change for this column
        mean_change = diffs[column].mean()
        mean_changes[column] = mean_change
        
        # Calculate standard deviation
        std_change = diffs[column].std()
        
        # Calculate z-scores
        if std_change > 0:
            z_scores = (diffs[column] - mean_change) / std_change
        else:
            z_scores = pd.Series(0, index=diffs.index)
        
        # Find days with significant changes (|z| > 1.0)
        sig_days = diffs.index[abs(z_scores) > 1.0]
        
        if len(sig_days) > 0:
            # Create a DataFrame for significant days
            significant_days_df = pd.DataFrame(index=sig_days)
            significant_days_df[column] = diffs.loc[sig_days, column]
            # Add z-scores to the significant days
            significant_days_df['z_score'] = z_scores.loc[sig_days]
            
            # Store in dictionary
            significant_days[column] = significant_days_df
    
    return significant_days, mean_changes

sig_days_1, mean_changes_1 = detect_significant_days(diffs_1)
sig_days_2, mean_changes_2 = detect_significant_days(diffs_2)
sig_days_3, mean_changes_3 = detect_significant_days(diffs_3)


# Create summary DataFrames for each experiment
summary_dfs = []

for exp_num, (means, diffs, sig_days, mean_changes) in enumerate([
    (means_1, diffs_1, sig_days_1, mean_changes_1),
    (means_2, diffs_2, sig_days_2, mean_changes_2),
    (means_3, diffs_3, sig_days_3, mean_changes_3)
], 1):
    
    # For each genotype-treatment combination
    for column in diffs.columns:
        genotype, treatment = column
        
        # Create a summary dataframe for this combination
        summary = pd.DataFrame()
        summary['experiment'] = [exp_num] * len(diffs.index)
        summary['day'] = diffs.index
        summary['genotype'] = genotype
        summary['treatment'] = treatment
        summary['mean_transpiration'] = means[column].loc[diffs.index].values
        summary['daily_change'] = diffs[column].values
        summary['avg_change'] = mean_changes[column]
        
        # Calculate z-scores
        std_change = diffs[column].std()
        z_scores = (diffs[column] - mean_changes[column]) / std_change if std_change > 0 else 0
        summary['z_score'] = z_scores
        summary['significant'] = abs(z_scores) > 2.0
        
        summary_dfs.append(summary)

# Combine all summaries
all_summaries = pd.concat(summary_dfs).reset_index(drop=True)
print("\nSummary DataFrame (first few rows):")
print(all_summaries)
```

```python
# Extract rows where 'significant' is True
significant_rows = all_summaries[all_summaries['significant'] == True]

# Display the result
print("Rows with significant changes:")
print(significant_rows)

# If you want to know how many significant days were found
print(f"\nFound {len(significant_rows)} days with significant changes")
```
